```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Load dataset
data = pd.read_csv("electricity.csv")

# Create line plot
sns.lineplot(
    x="Date",
    y="Electricity Consumption",
    hue="Consumer Category",
    linestyle="--",
    style="Consumer Category",
    marker="o",
    data=data
)

# Add labels and title
plt.xlabel("Time (Months)")
plt.ylabel("Electricity Consumption (kWh)")
plt.title("Monthly Electricity Consumption Trend")

# Customize X-axis
plt.xticks(
    rotation=45,
    ha="center",
    fontsize=12,
    fontfamily="serif",
    fontweight="bold"
)

# Customize Y-axis
plt.yticks(
    rotation=45,
    ha="center",
    fontsize=12,
    fontfamily="serif",
    fontweight="bold"
)

# Add grid
plt.grid(True)

# Display plot
plt.show()
```
